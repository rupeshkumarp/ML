```python
# Binary Search

arr = [10, 20, 30, 40, 50, 60, 70]   # Sorted list
key = int(input("Enter the number to search: "))

low = 0
high = len(arr) - 1

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        print(f"Element found at index {mid}")
        break
    elif arr[mid] < key:
        low = mid + 1
    else:
        high = mid - 1
else:
    print("Element not found")
```
