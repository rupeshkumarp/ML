```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Jeevan kumar\Downloads\MathsnStats_normal_distribution_resources\MathsnStats_normal_distribution_resources\heights.csv")
df.head()
```

```python
df.height.plot(kind='box')
```

```python
mean,std=round(df['height'].mean(),2),round(df['height'].std(),2)
mean,std
```

```python
# df.height.plot(kind='hist',edgecolor='black',bins=15)
sns.histplot(df.height,kde=True)
plt.axvline(mean,color='b',linestyle='-')

plt.axvline(mean+std,color='r',linestyle='--')
plt.axvline(mean-std,color='r',linestyle='--')

plt.axvline(mean+2*std,color='r',linestyle='--')
plt.axvline(mean-2*std,color='r',linestyle='--')

plt.axvline(mean+3*std,color='r',linestyle='--')
plt.axvline(mean-3*std,color='r',linestyle='--')
```

```python
left1=mean-std
right1=mean+std

left2,right2=round(mean-2*std,2),round(mean+2*std,2)

left3,right3=round(mean-3*std,2),round(mean+3*std,2)
```

```python
left1,left2,left3
```

```python
right1,right2,right3
```

```python
df[(df['height']>left1)&(df['height']<right1)].shape #65% of data
```

```python
df[(df['height']>left2)&(df['height']<right2)].shape #97% of data
```

```python
df[(df['height']>left3)&(df['height']<right3)].shape #only outlier have leaven out
```

```python
def get_low_upper(data):
    q1,q3=data['height'].quantile([.25,.75])
    iqr=q3-q1
    low=round(q1-1.5*iqr,2)
    up=round(q3+1.5*iqr,2)
    return low,up
```

```python
low,upp=get_low_upper(df)
```

```python
df[(df['height']<low)|(df['height']>upp)]
```

```python
df['z_score']=(df['height']-mean)/std
```

```python
df[(df['z_score']<-3)|(df['z_score']>3)]
```

```python
sns.histplot(df.z_score,kde=True) #Standard Normal Distribution
```

```python
outliers = df[(df['height'] < low) | (df['height'] > upp)]
```

```python
df=df.drop(outliers.index)
```
