```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df.isnull().sum()
```

```python
df.nunique()
```

```python
df1=df[['Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount', 
               'Distance', 'CouncilArea', 'Bedroom2', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Price']]
```

```python
df1.sample(5)
```

```python
df['Propertycount']=df['Propertycount'].fillna(0)
df['Distance']=df['Distance'].fillna(0)
df['Bedroom2']=df['Bedroom2'].fillna(0)
df['Bathroom']=df['Bathroom'].fillna(0)
df['Car']=df['Car'].fillna(0)
```

```python
df1['Landsize'] = df1['Landsize'].fillna(df1['Landsize'].mean())
```

```python
df1['BuildingArea']=df['BuildingArea'].fillna(df1['BuildingArea'].mean())
```

```python
df.isnull().sum()
```

```python
df1.dropna(inplace=True)
```

```python
data_set=pd.get_dummies(df1, drop_first=True)
data_set.sample(5)
```

```python
data_set.shape
```

```python
x=data_set.drop('Price', axis=1)
y=data_set['Price']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3,random_state=2)
```

```python
model=LinearRegression()
model.fit(x_train, y_train)
```

```python
model.score(x_test,y_test)
```

```python
from sklearn import linear_model
lesso=linear_model.Lasso(alpha=50,max_iter=100,tol=0.1)
lesso.fit(x_train,y_train)
```

```python
lesso.score(x_test,y_test)
```

```python
from sklearn.linear_model import  Ridge  
ridge=Ridge(alpha=50,max_iter=100,tol=0.1)
ridge.fit(x_train,y_train)
```

```python
ridge.score(x_test,y_test)
```
