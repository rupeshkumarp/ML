```python
sample_size = 30

sample=df.sample(sample_size,random_state=100)
sample.head()   
```

```python
sample_mean,sample_std=sample.miles.mean(),sample.miles.std()
sample_mean,sample_std
```

```python
mean_p=df.miles.mean()
```

```python
std_p=df.miles.std()
```

```python
#95 confedence interval
import scipy.stats as st
z_score_95=st.norm.ppf(0.975)
z_score_95
```

```python
margin_of_error=z_score_95 * (std_p/np.sqrt(sample_size))
margin_of_error
```

```python
lower=sample_mean-margin_of_error
upper=sample_mean+margin_of_error
lower, upper
```

```python
z_score_99=st.norm.ppf(0.995)
z_score_99
```

```python
margin_of_error_99=z_score_99 * (std_p/np.sqrt(sample_size))
margin_of_error_99
```

```python
lower_limit_99=sample_mean-margin_of_error_99
upper_limit_99=sample_mean+margin_of_error_99
lower_limit_99, upper_limit_99
```
