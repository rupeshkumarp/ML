```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\Jeevan kumar\Downloads\movie_revenues.csv")
df.head(2)
```

```python
df.revenue.describe()
```

```python
df['revenue_millions']=df.revenue/1000000
df.head(2)
```

```python
df['modified_z_score']=0.6745*(df.revenue-df.revenue.median())/df.revenue.mad()
df.head(2)
```

```python
_,mean,std,*_=df.revenue_millions.describe()
mean,std
```

```python
def get_z_score(x,mean,std):
    return (x-mean)/std
```

```python
df['zscore']=df.revenue_millions.apply(lambda x:get_z_score(x,mean,std))
```

```python
df[df['zscore']>3.5]
```

```python
def get_mad(s):
    median=np.median(s)
    deff=abs(s-median)
    MAD=np.median(deff)
    return MAD
```

```python
mad=get_mad(df.revenue_millions)
meadian=np.median(df.revenue_millions)
mad,meadian
```

```python
def get_modified_z_score(x,median,mad):
    return 0.6745*(x-median)/mad 
```

```python
df['modified_z_score']=df.revenue_millions.apply(lambda x:get_modified_z_score(x,meadian,mad))
```

```python
df[df['modified_z_score']>3.5]
```
