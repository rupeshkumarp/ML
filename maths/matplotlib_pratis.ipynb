```python
import pandas as pd
```

```python
df=pd.DataFrame(data)
```

```python
df["age"]=[23,34,22,24,25,27,45,35,24,43,43,35,32,21,25,43]
```

```python
sort_age=df.sort_values('age')
```

```python
plt.scatter(sort_age['age'],df['salary'],marker='*')
plt.show
```

```python
plt.plot(sort_age['age'],df['salary'],linestyle="-",marker='*',color='red')
plt.show
```

```python
plt.bar(sort_age['age'],df['salary'],color='red',align='edge')
plt.figure(figsize=(5,15))
plt.show
```

```python
hr_sal=df[df['dept']=="HR"]['salary']
hr_sal
```

```python
it_sal=df[df['dept']=='IT']['salary']
it_sal
```

```python
fin_sal=df[df['dept']=='finance']['salary']
dev_sal=df[df['dept']=='Developer']['salary']
```

```python
plt.boxplot([hr_sal,it_sal,fin_sal,dev_sal],labels=['hr_sal','it_sal','fin_sal','dev_sal'])
plt.legend(['HR','IT','Finance','Developer'])
plt.grid()
plt.show
```

```python
sal_dept=df.groupby('dept')['salary'].sum()
sal_dept
```

```python
plt.pie(sal_dept,labels=sal_dept.index,autopct='%1.1f',shadow=True,explode=[0,0,0.1,0],wedgeprops={'edgecolor':'black','linewidth': 1})
plt.grid()
plt.show()
```

```python
mean_sal_dept=df.groupby('dept')['salary'].mean().round(2)
mean_sal_dept
```

```python
plt.bar(['DEVELOPER','HR','IT','FINANCE'],mean_sal_dept,color=['red','blue','green','yellow'])
plt.grid()
plt.show()
```

```python
df['experience']=[4,6,2,4,5,3,4,5,3,2,5,5,3,4,4,8]
```

```python
df.groupby(['dept','experience'])['salary'].mean()
```

```python
plt.scatter(df['age'],df['salary'],df['experience']*20,color='green',marker='o',edgecolors='black')
plt.grid()
plt.show()
```

```python
colour={'HR':'yellow','Developer':'red','finance':'blue','IT':'black'}
for dept ,color in colour.items():
    df_dept=df[df['dept']==dept]
    plt.scatter(df_dept['age'],df_dept['salary'],c=color,label=dept,marker='*')
plt.legend()
```

```python
fig, axs=plt.subplots(1,4, figsize=(20,5))
axs[0].plot(sort_age['age'],df['salary'],marker='*',color='blue',linestyle=':',markersize=3,label='age')
axs[0].set_xlabel('age')
axs[0].set_ylabel('salary')
axs[0].legend()
axs[0].grid()

axs[1].hist(df['salary'],bins=5,edgecolor='black')
axs[1].set_xlabel('salary')
axs[1].set_ylabel('frequency')
axs[1].grid()

axs[2].boxplot(df['salary'])
axs[2].set_xlabel('salary')
axs[2].set_ylabel('frequency')
axs[2].grid()

axs[3].scatter(sort_age['age'],df['experience'],marker='*')
axs[3].grid()

plt.tight_layout()
plt.show()
```

```python
ax=plt.axes(projection='3d')
ax.scatter(sort_age['age'],df['salary'],df['experience'],marker='*')
ax.grid()
ax.set_xlabel('age')
ax.set_ylabel('salary')
ax.set_zlabel('experience')
ax.view_init(elev=20, azim=45)


plt.show()
```

```python
fig = plt.figure(figsize=(8,6))
ax = plt.axes(projection='3d')

ax.hist(df['salary'],bins=4)

ax.set_xlabel('Age')
ax.set_ylabel('Salary')
ax.set_title("3D Scatter Plot: Age vs Salary vs Experience")

ax.grid()
plt.show()
```
