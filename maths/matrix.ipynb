```python
import numpy as np
```

```python
v1=np.array([[1, 2, 3],[4,5,6]])
v2=np.array([[7, 8, 9],[10,11,12]])
```

```python
np.percentile(v1,99)
```

```python
v1+v2
```

```python
np.dot(v1,v2)
```

```python
v1.T
```

```python
np.linalg.det([[1, 2], [3, 4]])
```

```python
np.linalg.eig([[1, 2], [3, 4]])
```

```python
a=np.array([[1,2],[3,5]])
a
```

```python
np.linalg.inv(a)
```

```python
np.linalg.eig(a)
```

```python
np.linalg.solve(a,[[6,7],[8,9]])
```

```python
np.linalg.svd(a)
```
