```python
g = df.groupby("City")
sample_means = []
for city, subdf in g:
    sample_means.append(subdf.Bill.mean())
sample_means[:5]
```

```python
sns.histplot(sample_means, kde=True)

plt.xlabel("Electricity Bill Sample Means")
plt.show()
```

```python
sample_m=pd.DataFrame(sample_means, columns=['Sample Means'])
sample_m.describe()
```

```python
mean=sample_m['Sample Means'].mean()
```

```python
sns.histplot(sample_m['Sample Means'], kde=True)
plt.axvline(mean, color='red',linestyle='--', label='Mean of Sample Means')
```
