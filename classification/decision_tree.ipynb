<h1>feature engineering</h1>

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['degree'] = le.fit_transform(df['degree'])
```

```python
from sklearn.model_selection import train_test_split
```

<h1>Model training</h1>

```python
from sklearn.tree import DecisionTreeClassifier

x=df_encoded.drop('salary_more_then_100k',axis='columns')
y=df_encoded['salary_more_then_100k']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=5)
model=DecisionTreeClassifier() # deafult gimi
model.fit(x,y)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
y_pred=model.predict(x_test)

report=classification_report(y_test,y_pred)
print(report)
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

<h1>Decision tree</h1>

```python
from sklearn.tree import export_text

print(export_text(model,feature_names=list(x.columns)))
```

<h1>Entropy</h1>

```python
from sklearn.tree import DecisionTreeClassifier

x=df_encoded.drop('salary_more_then_100k',axis='columns')
y=df_encoded['salary_more_then_100k']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=5)
model = DecisionTreeClassifier(
    criterion='gini',#how the tree evaluates the quality of a split.
    max_depth=5,# depth
    min_samples_split=10,#minimum number of samples required in a node before that node can be split.
    min_samples_leaf=5,#minimum number of samples that must remain in each leaf.
    max_features=None,#how many features the tree considers when looking for a split.
    random_state=42#controls randomness so you can get reproducible results.
)
model.fit(x,y)
```

```python
y_pred=model.predict(x_test)

report=classification_report(y_test,y_pred)
print(report)
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
model.score(x_test,y_test)
```
