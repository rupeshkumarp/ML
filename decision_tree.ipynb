```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Jeevan kumar\Downloads\salaries.csv")
```

```python
df.head()
```

```python
input=df[['company','job','degree']]
target=df['salary_more_then_100k']
```

```python
input.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le_company=LabelEncoder()
le_job=LabelEncoder()
le_degree=LabelEncoder()
#create new columns for encoded values (categorical to numerical)
```

```python
input['company_n']=le_company.fit_transform(input['company'])
input['job_n']=le_job.fit_transform(input['job'])
input['degree_n']=le_degree.fit_transform(input['degree'])
```

```python
input.drop(['company','job','degree'],axis='columns',inplace=True)
```

```python
input.sample(5)
```

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
```

```python
model.fit(input,target)
```

```python
model.predict([[1,0,2]])
```

```python
model.predict([[2,0,0]])
```
