```python
df=pd.read_csv(r"C:\Users\Jeevan kumar\Downloads\spam.csv")
df.sample(5)
```

```python
df.groupby('Category').describe()
```

```python
df.Category.value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Category']=le.fit_transform(df['Category'])
df.head()
```

```python
from sklearn.model_selection import train_test_split
x=df['Message']
y=df['Category']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
v=CountVectorizer()
X_train=v.fit_transform(x_train)
X_train.toarray()
```

```python
from sklearn.naive_bayes import MultinomialNB
model_nv=MultinomialNB()
```

```python
model_nv.fit(X_train,y_train)
```

```python
model_nv.score(v.transform(x_test),y_test)
```

```python
e=["Congratulations! You've won a $1,000 Walmart gift card. Click here to claim your prize.",
   "Hey, are we still meeting for lunch today?"]
model_nv.predict(v.transform(e))
```

```python
from sklearn.pipeline import Pipeline
clf=Pipeline([('vectorizer',CountVectorizer()),
              ('naive_bayes',MultinomialNB())])
```

```python
clf.fit(x_train,y_train)
```

```python
clf.score(x_test,y_test)
```

```python
clf.predict(e)
```
