```python
df[df['height']<=upper_limit]
```

```python
df[(df['height']>upper_limit) | (df['height']<lower_limit)]
```

```python
df[(df['height']<upper_limit) & (df['height']>lower_limit)]
```

```python
outlier_removed=df[(df['height']<=upper_limit)& (df['height']>=lower_limit) ]
outlier_removed
```

```python
# df.height.replace([df.height>upper_limit,df.height<lower_limit],df.mean())
df['height'] = np.where(
    (df['height'] > upper_limit) |
    (df['height'] < lower_limit),
    df['height'].mean(),
    df['height']
)
```

```python
df_s=pd.read_excel(r"C:\Users\Jeevan kumar\Downloads\mathsnstats_iqr_boxplot_resources\7_outlier_detection_using_iqr_boxplot\region_wise_sales.xlsx")
df_s
```

```python
df_s.groupby('Region').agg({'Sales':'sum'}).plot(kind='box')
```

```python
df_s.Region.unique()
```

```python
df_APAC=df_s[df_s.Region=='APAC']
df_Europe=df_s[df_s.Region=='Europe']
df_Americas=df_s[df_s.Region=='Americas']
```

```python
def get_lower_upper(data):
    q1, q3 = data['Sales'].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return lower, upper
```

```python
l,u=get_lower_upper(df_APAC)
l,u
```

```python
la,ua=get_lower_upper(df_Americas)
la,ua
```

```python
le,ue=get_lower_upper(df_Europe)
le,ue
```

```python
import matplotlib.pyplot as plt
```

```python
df_s.boxplot(column='Sales',by='Region')
```

```python
df_Europe[df_Europe.Sales>ue]
```

```python
labels=df_s['Region'].unique()
plot_data=[df_s['Sales'][df_s['Region']==label].values for label in labels]
plot_data
```

```python
plt.boxplot(plot_data,labels=labels,vert=True,patch_artist=True)
plt.xticks(rotation=45)
plt.show()
```
