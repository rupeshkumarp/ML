# Requirements and Scope of Work (SOW)

<p>
1.A SOW document is a mix of business
requirement and the technology being used.
This acts as a reference point for all
stakeholders involved.<br>
2. Project estimation needs to be done with a
consideration that in the initial phase the
requirements are vague, and the timeline
estimate typically gets better as the time
passes.<br>
3.JIRA is a popular software used in the industry
to manage AI/ML projects.<br>
4.Other popular tools used for project
management are confluence, Notion, Asana,
Excel etc.</p>
