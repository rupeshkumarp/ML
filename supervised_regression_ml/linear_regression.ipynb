<h1>polynomial regression

```python
df=pd.read_csv(r"C:\Users\Jeevan kumar\Videos\dataset\car_prices.csv")
df.sample(5)
```

```python
df.dtypes
```

```python
plt.scatter(df.mileage,df.selling_price)
```

```python
x=df[['mileage']]
y=df.selling_price
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
x_train[:5]
```

```python
x_poly[:5]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
model=LinearRegression()
model.fit(x_poly,y_train)
```

```python
y_pred=model.predict(x_test_poly)
```

```python
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
mse,r2
```

```python
x_range=np.linspace(x['mileage'].min(),x['mileage'].max(),300).reshape(-1,1)
x_range_poly=poly.transform(x_range)
y_range=model.predict(x_range_poly)
```

```python
plt.scatter(x_test['mileage'],y_test,color='b',label='Actual')
plt.plot(x_range[:, 0], y_range, color='r', label='Predicted')
plt.xlabel('Mileage')
plt.ylabel('Selling Price')
plt.title('Linear Regression')
plt.legend()
plt.show()
```
