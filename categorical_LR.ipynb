```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Jeevan kumar\Downloads\homeprices.csv")
```

```python
df.head()
```

```python
dummies = pd.get_dummies(df['town']).astype(int)
```

```python
dummies
```

```python
merged = pd.concat([df,dummies],axis='columns')
```

```python
merged.head()
```

```python
merged.drop(['town','west windsor'],axis='columns',inplace=True)
```

```python
merged
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
x=merged.drop('price',axis='columns')
y=merged.price
```

```python
model.fit(x,y)
```

```python
model.predict([[2800,1,0]])
```

```python
model.predict([[3400,0,0]])
```

```python
model.score(x,y)
```
